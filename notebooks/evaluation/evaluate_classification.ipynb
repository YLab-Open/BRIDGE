# Classification Evaluation Example

This notebook loads `evaluation/evaluate_BRIDGE.yaml`, narrows `task_types` to `clf`, and keeps execution behind `RUN_EVALUATION`. Edit the YAML file for models, prompt modes, result paths, and task subsets.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
for candidate in [PROJECT_ROOT, *PROJECT_ROOT.parents]:
    if (candidate / "main.py").exists():
        PROJECT_ROOT = candidate
        break
else:
    PROJECT_ROOT = Path.cwd().resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(PROJECT_ROOT)
```

```python
from evaluation.bridge import config_from_yaml, evaluate_all
from evaluation.result_utils import collect_result_inventory

path_config = PROJECT_ROOT / "evaluation" / "evaluate_BRIDGE.yaml"
config = config_from_yaml(path_config)
config.task_types = ["clf"]
config.print_summary = False
config
```

```python
inventory = collect_result_inventory(
    dir_result=config.dir_result,
    task_names=config.tasks,
    model_names=config.model_names,
    prompt_modes=config.prompt_modes,
    count_rows=False,
)
inventory[:10]
```

```python
RUN_EVALUATION = False
if RUN_EVALUATION:
    evaluate_all(config)
else:
    print("Dry run only. Set RUN_EVALUATION = True to evaluate and write performance JSON files.")
```
